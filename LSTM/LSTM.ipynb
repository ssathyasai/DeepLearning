```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM,Dense
```

```python
#Input Sentence
sentence=["I","Love","Deep","Learning"]
print(sentence)
```

```python
#Wrod Embeddingd
word_embeddings={
    "I":1.0,
    "Love":2.0,
    "Deep":3.0,
    "Learning":4.0
}
print(word_embeddings)
```

```python
#Prepare Training Data
X=np.array([
    [
        [word_embeddings["I"]],
        [word_embeddings["Love"]],
        [word_embeddings["Deep"]]
    ]

])
print(X)
```

```python
y=np.array([
    [word_embeddings["Learning"]]
])
print(y)
```

```python
print("Input Shape:",X.shape)
print("Output Shape:",y.shape)
```

```python
#Build LSTM Model
model = Sequential()
model.add(
    LSTM(
        units=8,
        activation='tanh',
        input_shape=(3,1)
        )
)
model.add(Dense(1))
model.summary()
```

```python
#Compile Model
model.compile(
    optimizer='adam',
    loss='mse'
)
```

```python
#Train the Model
model.fit(X,y,epochs=300,verbose=1)
```

```python
#Prediction
test_sequence=np.array([
    [
        [1.0],
        [2.0],
        [3.0]
    ]
])
prediction=model.predict(test_sequence)
print(prediction)
#
```

```python
#Convert prediction to Word
predict_number=prediction[0][0]
closest_word=min(
    word_embeddings.keys(),
    key=lambda word:abs(word_embeddings[word]-predict_number)
)
print("Predicted Word:",closest_word)
```
